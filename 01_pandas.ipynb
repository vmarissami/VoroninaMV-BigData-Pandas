# Практическая работа 1. Выборка данных в pandas

Вариант 8

```python
import pandas as pd

df = pd.DataFrame({
    "ID": range(1, 11),
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank", "Grace", "Helen", "Ivan", "Julia"],
    "Age": [23, 35, 29, 40, 28, 45, 32, 27, 38, 31],
    "City": ["Moscow", "SPb", "Moscow", "Kazan", "SPb", "Moscow", "SPb", "Kazan", "Moscow", "SPb"],
    "Salary": [50000, 70000, 60000, 80000, 55000, 90000, 65000, 58000, 75000, 62000],
})
```

### 1. Первые пять строк

```python
df.head()
```

### 2. Строки с чётными индексами

```python
df.loc[df.index % 2 == 0]
```

### 3. Имена длиннее пяти символов

```python
df.loc[df["Name"].str.len() > 5]
```

### 4. Средний возраст

```python
print(f'Средний возраст — {df["Age"].mean():.1f} года.')
```

### 5. Сотрудники от 35 лет

```python
df["Senior"] = df["Age"] >= 35
df[["Name", "Age", "Senior"]]
```
