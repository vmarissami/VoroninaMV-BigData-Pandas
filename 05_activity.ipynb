```python
import pandas as pd
from shop_data import read_events, read_orders

orders = read_orders()
events = read_events()
```

### 1–2. Признаки присутствия

Список пользователей берём из объединения двух источников. Так в таблицу попадёт и человек, который только просматривал товары, и покупатель без записей в журнале.

```python
event_users = pd.Index(events["user_id"].unique())
order_users = pd.Index(orders["user_id"].unique())
users = event_users.union(order_users).sort_values()
activity = pd.DataFrame({
    "has_events": users.isin(event_users),
    "has_orders": users.isin(order_users),
}, index=users.rename("user_id"))
activity.head()
```

### 3. Сколько пользователей есть в каждом источнике

```python
counts = pd.crosstab(activity["has_events"], activity["has_orders"])
counts = counts.reindex(index=[False, True], columns=[False, True], fill_value=0)
counts
```

Все 100 наблюдаемых пользователей встречаются и в логах, и в заказах. Группы «только события» и «только заказы» пусты. Число людей, которые вообще не попали ни в один файл, по этим данным определить нельзя. Совпадение по `user_id` также не показывает, после какого именно события появился заказ.
